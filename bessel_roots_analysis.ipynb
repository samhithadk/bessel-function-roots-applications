```python
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
from scipy.integrate import quad
from scipy.optimize import fsolve
```

```python
# make function
def f_1(theta, x, n):
  return np.cos(x*np.sin(theta) - n * theta)

def f_2(x, n):
  return (1/np.pi) * quad(f_1, 0, np.pi, args = (x, n))[0]

x_values = np.linspace(0, 20, 200)

# get values for when n is 0 across x
n0 = np.array([f_2(x, 0) for x in x_values])

# get values for when n is 1 across x
n1 = np.array([f_2(x, 1) for x in x_values])

# get values for when n is 2 across x
n2 = np.array([f_2(x, 2) for x in x_values])

# make plot to show all values of x
plt.plot(x_values, n0, label = 'n = 0')
plt.plot(x_values, n1, label = 'n = 1')
plt.plot(x_values, n2, label = 'n = 2')
plt.legend()
plt.grid()
plt.xlabel('x')
plt.ylabel('Bessel Function')
plt.hlines(0, 0, 20, linestyles = 'dashed', color = "black")
plt.show()

```

```python
# approx roots
n0_roots_approx = np.array([2.4, 5.2, 8.7, 11.5, 14.8])
n1_roots_approx = np.array([3.5, 7, 10.1, 13.5, 16.7])
n2_roots_approx = np.array([5, 8.2, 11.8, 14.9, 18])

# finding the roots based on approx guess using a for loop
n0_roots = [fsolve(f_2, approx_root, args = (0)) for approx_root in n0_roots_approx]
n1_roots = [fsolve(f_2, approx_root, args = (1)) for approx_root in n1_roots_approx]
n2_roots = [fsolve(f_2, approx_root, args = (2)) for approx_root in n2_roots_approx]

print("Roots for n=0:", n0_roots)
print("Roots for n=1:", n1_roots)
print("Roots for n=2:", n2_roots)
```

```python
# plot roots as points
plt.hlines(0, 0, 20, linestyles = 'dashed', color = "black")
plt.plot(n0_roots, np.zeros(5), 'o', label = '_n=0', color = '#1f77b4')
plt.plot(n1_roots, np.zeros(5), 'o', label = '_n=1', color = '#ff7f0e')
plt.plot(n2_roots, np.zeros(5), 'o', label = '_n=2', color = '#2ca02c')
plt.plot(x_values, n0, label = 'n = 0', color = '#1f77b4')
plt.plot(x_values, n1, label = 'n = 1', color = '#ff7f0e')
plt.plot(x_values, n2, label = 'n = 2', color = '#2ca02c')
plt.legend()
plt.grid()
plt.xlabel('x')
plt.ylabel('Bessel Function')
plt.show()
```
